```python
# ==============================================================================
# RED NEURONAL QUE APRENDE A CONVERTIR DE CELSIUS A FAHRENHEIT
# ==============================================================================

# 'import' es la palabra clave para CARGAR librerías externas en nuestro programa.
# 'tensorflow' es la librería principal desarrollada por Google para crear redes neuronales.
# 'as tf' le asigna un alias corto ('tf') para facilitar su lectura en el código.
import tensorflow as tf

# 'numpy' es la librería matemática esencial para gestionar listas numéricas (arreglos) con alta velocidad.
import numpy as np


# ------------------------------------------------------------------------------
# 1. PREPARACIÓN DE LOS DATOS DE ENTRENAMIENTO
# ------------------------------------------------------------------------------

# 'np.array' crea un arreglo numérico optimizado con las entradas (Celsius).
# 'dtype=float' especifica explícitamente que los números contienen decimales.
celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)

# Creación del arreglo con las respuestas reales esperadas (Fahrenheit).
fahrenheit = np.array([-40, 14, 32, 46.4, 59, 71.6, 100.4], dtype=float)


# ------------------------------------------------------------------------------
# 2. DISEÑO Y CONSTRUCCIÓN DE LA RED NEURONAL
# ------------------------------------------------------------------------------

# Definimos explícitamente la entrada del modelo usando un objeto Input para evitar warnings.
capa_entrada = tf.keras.layers.Input(shape=(1,))

# 'tf.keras.layers.Dense' crea una capa de neuronas densamente conectada.
# 'units=1' define que esta capa tiene una sola neurona.
capa_salida = tf.keras.layers.Dense(units=1)

# 'tf.keras.Sequential' es la palabra clave/clase para APILAR capas en orden secuencial.
modelo = tf.keras.Sequential([capa_entrada, capa_salida])


# ------------------------------------------------------------------------------
# 3. COMPILACIÓN DEL MODELO (Ajuste de reglas de aprendizaje)
# ------------------------------------------------------------------------------

# '.compile()' prepara la red asignándole un optimizador y una medida de error antes de entrenar.
modelo.compile(
    # 'optimizer' especifica la estrategia matemática para ir ajustando las perillas (pesos).
    # 'Adam(0.1)' es un algoritmo de ajuste muy eficiente; '0.1' es la velocidad a la que aprende (tasa de aprendizaje).
    optimizer=tf.keras.optimizers.Adam(0.1),
    # 'loss' define la función para MEDIR EL ERROR.
    # 'mean_squared_error' (error cuadrático medio) evalúa qué tan lejos estuvo la predicción de la realidad.
    loss="mean_squared_error",
)


# ------------------------------------------------------------------------------
# 4. ENTRENAMIENTO DE LA RED NEURONAL
# ------------------------------------------------------------------------------

# 'print()' muestra un mensaje descriptivo en la consola.
print("🧠 Entrenando la red neuronal... Por favor espera un momento.")

# '.fit()' es la instrucción que INICIA EL PROCESO DE ENTRENAMIENTO.
# 'celsius' representa las entradas y 'fahrenheit' las respuestas correctas.
# 'epochs=500' indica que la red repasará todos los datos 500 veces para ajustar sus parámetros.
# 'verbose=False' oculta el detalle detallado de cada ciclo para mantener limpia la consola.
historial = modelo.fit(celsius, fahrenheit, epochs=500, verbose=False)

print("¡Entrenamiento completado con éxito!\n")


# ------------------------------------------------------------------------------
# 5. VERIFICACIÓN DE PREDICCIONES Y PARÁMETROS APRENDIDOS
# ------------------------------------------------------------------------------

# Definimos una temperatura de prueba en Celsius que la red evaluará.
valor_prueba = 100.0

# Convertimos el valor de prueba a un arreglo de NumPy compatible con el método predict.
prediccion_fahrenheit = modelo.predict(np.array([valor_prueba]))

# Mostramos el resultado predicho por la inteligencia artificial.
print(
    f"Convertido a Fahrenheit: {prediccion_fahrenheit[0][0]:.2f}°F"
)
print(f"📊 Valor real según la fórmula matemática: 212.00°F\n")

# '.get_weights()' extrae los valores internos aprendidos automáticamente por la neurona.
pesos = capa_salida.get_weights()

# 'pesos[0]' representa el 'peso' (que debería aproximarse a 1.8, equivalente a 9/5).
print(f"Peso aprendido por la neurona (Multiplicador): {pesos[0][0][0]:.4f}")

# 'pesos[1]' representa el 'sesgo' o bias (que debería aproximarse a 32).
print(f"Sesgo aprendido por la neurona (Suma constante): {pesos[1][0]:.4f}")
```
