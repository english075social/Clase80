# Explicacion gemini
# Solicitamos al usuario un número entero
numero = int(input("Introduce un número entero: "))

# '%' es el operador módulo, calcula el residuo de dividir 'numero' entre 2
# '==' es el operador de comparación que verifica si dos valores son exactamente iguales
if numero % 2 == 0:
    # Bloque de código que se ejecuta solo si la condición es Verdadera (True)
    print(f"El número {numero} es PAR.")
else:
    # 'else:' marca el camino alternativo si la condición resultó Falsa (False)
    print(f"El número {numero} es IMPAR.")
