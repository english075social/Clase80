A traves de este concepto aprendeemos  los conceptos básicos detras de la regresión lineal y como implementarla en python unando biblioteccas como: Pandas, numpy, Metplolib, Sckit-learn

```python
#importamos librerias
import pandas as pd #manipulacion de datos
import numpy as np #para calculos numericos
import matplotlib.pyplot as plt #para visualizacion
from sklearn.linear_model import LinearRegression # para modelado de la reg....
import warnings # para gestionar las adcertencias

warnings.filterwarnings('ignore') # ignorar las advertencias
#Leer el archivo CSV "Salary.csv" y cargar los datos de un dataframe
df = pd.read_csv('Salary.csv')
#mostrar las primeras 6 filas del DataFrame
df.head(6)
```

```python
#Grafiquemos los datos para observar cual es la correlación entre las variables
#Crear un grafico de dispersión
plt.scatter(df["YearsExperience"], df["Salary"], color='blue')

#Agregar etiquetas al eje x y y
plt.xlabel("Años de Experiencia")
plt.ylabel("Salario")
plt.title("Relación de Experiencia vs Salario")

#mostrar el grafico
plt.show()
```

```python
# Cear un objeto de modelo de regresion lineal
modelo= LinearRegression()
#Ajustamos el modelo utilizando los datos de la tabla
modelo.fit(df[["YearsExperience"]], df["Salary"])

#realizamos una prediccion a 15 años
prediccion = modelo.predict([[15]])
#mostrar la prediccion

#imprimir la prediccion
print(f"Prediccion del salario anual para una persona de 15 años de experiencia: {prediccion}")
```

```python
plt.scatter(df["YearsExperience"], df["Salary"]) # graico de dispersión
plt.plot(df["YearsExperience"], modelo.predict(df[["YearsExperience"]]), color='red')
#Supoerponer la linea de refresión del modelo en el gráfico

#Agregar etiquetas al eje x e y
plt.xlabel("Años de Experiencia")
plt.ylabel("Salario")
plt.title("Relación de Experiencia vs Salario")

#mostramos grafico de nuevo
plt.show()
```

```python
#Usaremos una metrica de evaluuación como coeficiente de  determinación (R^2)
# que nos permite medir la calidad de nuestras prediciones

#imprimir el coeficionet de determinación
print("R^2: {}".format(modelo.score(df[["YearsExperience"]], df["Salary"])))
```

Si el resultado de R^2 Es cercano a 1, es bastante bueno.
