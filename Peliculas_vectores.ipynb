```python
# 'import' carga librerías externas. 'numpy' es ideal para operaciones matemáticas y algebraicas.
# 'as np' le crea un apodo corto para no escribir 'numpy' a cada momento.
import numpy as np

# 1. DEFINICIÓN DE DATOS

# 'LISTA' de cadenas de texto (str) que almacena los géneros disponibles en un orden fijo.
GENEROS_COMPLETO = ["Acción", "Comedia", "Ciencia ficción", "Drama", "Suspense"]

# 'DICCIONARIO' principal donde cada 'clave' es el título de una película y su 'valor' es otro diccionario.
# Convertimos las listas binarias [1, 0, 1...] en 'np.array' para operar vectorialmente con NumPy.
PELICULAS_COMPLETO = {
    "Matrix": {
        "vector": np.array([1, 0, 1, 0, 1]), # 'np.array' crea un vector matemático eficiente.
        "tags": {"acción", "ciencia", "ficción", "futuro", "máquinas", "simulación", "hackeo", "computadoras"}
    },
    "El Padrino": {
        "vector": np.array([0, 0, 0, 1, 1]),
        "tags": {"drama", "mafia", "crimen", "familia", "clásicos", "obra", "maestra", "italia"}
    },
    "Toy Story": {
        "vector": np.array([0, 1, 0, 0, 0]),
        "tags": {"comedia", "animación", "infantil", "juguetes", "aventura", "amistad", "disney"}
    },
    "Gladiador": {
        "vector": np.array([1, 0, 0, 1, 0]),
        "tags": {"acción", "drama", "histórico", "roma", "imperio", "emperador", "venganza"}
    },
    "Superbad": {
        "vector": np.array([0, 1, 0, 0, 0]),
        "tags": {"comedia", "adolescentes", "escuela", "fiesta", "alcohol", "risas", "amigos"}
    },
    "Interestelar": {
        "vector": np.array([0, 0, 1, 1, 0]),
        "tags": {"ciencia", "ficción", "drama", "espacio", "viaje", "tiempo", "planetas", "agujero", "negro"}
    },
    "Duro de Matar": {
        "vector": np.array([1, 0, 0, 0, 1]),
        "tags": {"acción", "suspense", "policías", "terroristas", "rascacielos", "navidad", "disparos"}
    },
    "Parásitos": {
        "vector": np.array([0, 1, 0, 1, 1]),
        "tags": {"drama", "suspense", "comedia", "negra", "corea", "clases", "sociales", "intriga"}
    },
    "Inception": {
        "vector": np.array([1, 0, 1, 0, 1]),
        "tags": {"acción", "ciencia", "ficción", "suspense", "sueños", "subconsciente", "robos", "mente"}
    },
    "Pulp Fiction": {
        "vector": np.array([1, 0, 0, 1, 1]),
        "tags": {"crimen", "acción", "drama", "suspense", "comedia", "negra", "gángsters", "maletín"}
    },
    "El Club de la Pelea": {
        "vector": np.array([1, 0, 0, 1, 1]),
        "tags": {"acción", "drama", "suspense", "anarquía", "insomnio", "club", "peleas", "identidad"}
    }
}


# 2. FUNCIONES DE BÚSQUEDA Y CÁLCULO DE SIMILITUD

# 'def' declara una función reutilizable. Recibe el parámetro 'palabra_clave'.
def buscar_por_palabra(palabra_clave):
    # '.lower()' convierte el texto en minúsculas para evitar problemas de mayúsculas/minúsculas.
    palabra_clave = palabra_clave.lower().strip()

    # Lista vacía donde guardaremos los títulos coincidentes.
    coincidencias = []

    # 'for ... in ... .items()' recorre clave (titulo) y valor (info) del diccionario.
    for titulo, info in PELICULAS_COMPLETO.items():
        # 'if' evalúa si la palabra exacta existe dentro del conjunto de etiquetas ('tags').
        if palabra_clave in info["tags"]:
            # '.append()' añade el elemento al final de la lista.
            coincidencias.append(titulo)

    # 'return' devuelve el resultado de la función hacia donde fue llamada.
    return coincidencias


def calcular_similitud_unificada(peli1, peli2):
    # 'not in' verifica si alguna película no está registrada en el catálogo.
    if peli1 not in PELICULAS_COMPLETO or peli2 not in PELICULAS_COMPLETO:
        return 0

    # Extraemos los arreglos NumPy (vectores de 0s y 1s) correspondientes a cada película.
    vector1 = PELICULAS_COMPLETO[peli1]["vector"]
    vector2 = PELICULAS_COMPLETO[peli2]["vector"]

    # 'np.dot()' realiza el producto escalar (dot product).
    # Multiplica posición a posición y suma todos los resultados en una sola instrucción ultrarrápida.
    return np.dot(vector1, vector2)


# 3. INTERFAZ Y LÓGICA PRINCIPAL

def iniciar_recomendador():
    # 'print()' muestra mensajes formateados en la terminal.
    print("¡Bienvenido al Recomendador Unificado de Películas! 🍿\n")
    print("Pregunta 1: ¿Qué palabras o temáticas te interesan para hoy?")
    print("(Ejemplos: espacio, crimen, sueños, amistad, mafia, futuro, risas, roma)")

    # 'input()' pausa el programa esperando que el usuario escriba en la consola.
    # '.strip()' elimina espacios accidentales al inicio y final.
    palabra_clave = input("Escribe una palabra clave: ").strip().lower()

    # Ejecutamos la función de búsqueda.
    coincidencias = buscar_por_palabra(palabra_clave)

    # En Python, una lista con elementos se evalúa como Verdadera ('True').
    if coincidencias:
        print(f"\n¡Genial! Hemos encontrado estas opciones relacionadas: {coincidencias}")
        recomendada = coincidencias[0]
        print(f"Te sugerimos ver primero: **{recomendada}**")

        similares = []
        # '.keys()' obtiene todos los títulos de películas.
        for peli in PELICULAS_COMPLETO.keys():
            # '!=' significa 'diferente de' (evita comparar una película consigo misma).
            if peli != recomendada:
                puntos = calcular_similitud_unificada(recomendada, peli)
                # '>= 2' filtra aquellas con al menos 2 géneros en común.
                if puntos >= 2:
                    similares.append(peli)

        if similares:
            print(f"Si te gusta esta, también te podrían interesar: {similares}")
        return # 'return' interrumpe la función aquí porque ya encontramos respuesta.

    # Si no hubo coincidencias por palabra clave, pasa a la búsqueda por género.
    print("\nNo hemos encontrado películas con esa palabra clave exacta. Probemos con géneros.")
    print("Pregunta 2: Selecciona uno de los siguientes géneros:")

    # 'enumerate()' recorre la lista obteniendo tanto el índice (i) como el elemento (genero).
    for i, genero in enumerate(GENEROS_COMPLETO):
        print(f"{i + 1}. {genero}")

    # 'try-except' captura errores en tiempo de ejecución para evitar que el programa falle estrepitosamente.
    try:
        # 'int()' convierte la respuesta de texto ingresada por el usuario en un número entero.
        opcion_usuario = int(input("Escribe el número del género (1-5): "))

        # Restamos 1 porque la lista de Python empieza en el índice 0, pero al usuario le mostramos desde el 1.
        indice_genero = opcion_usuario - 1

        # Verificamos que el índice esté dentro del rango válido.
        if 0 <= indice_genero < len(GENEROS_COMPLETO):
            genero_elegido = GENEROS_COMPLETO[indice_genero]
            print(f"\nBuscando películas de: {genero_elegido}...")

            # Comprensión de listas: sintaxis compacta para crear una nueva lista filtrada.
            recomendadas = [
                titulo for titulo, info in PELICULAS_COMPLETO.items()
                if info["vector"][indice_genero] == 1
            ]

            if recomendadas:
                print(f"Películas del género {genero_elegido}: {recomendadas}")
            else:
                print("No se encontraron películas para este género.")
        else:
            print("Opción inválida. Debes seleccionar un número dentro del rango (1-5).")

    except ValueError:
        # Se ejecuta si el usuario escribe letras o símbolos en lugar de números.
        print("Error: Por favor, introduce un número entero válido.")

# Inicia la ejecución llamando a la función.
iniciar_recomendador()
```
