Clasificar la edad
Estructuras condicionales múltiples con elif

```python
edad = int(input("Introduce una edad verdadera: "))
if edad < 12:
  print("Eres un baby")
elif edad <= 17:
  print("eres un adololescente")
elif edad <= 64:
  print("eres un adulto")
else:
  print("eres un abulelo")
```

Eplicacion gemini
# Capturamos la edad ingresada como entero
edad = int(input("Introduce tu edad: "))

# 'if' evalúa la primera condición lógica
if edad < 12:
    print("Categoría: Niño/a")
# 'elif' es la abreviatura de 'else if'; se evalúa únicamente si la condición anterior fue Falsa
elif edad <= 17:
    print("Categoría: Adolescente")
elif edad <= 64:
    print("Categoría: Adulto")
else:
    # 'else' atrapa cualquier otro caso restante (en este contexto, 65 años o más)
    print("Categoría: Adulto Mayor")
