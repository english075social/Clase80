1. 🔍 Análisis Inicial y Diagnóstico
Genera una secuencia numérica ordenada y ejecuta una instrucción repetitiva por cada uno de sus elementos. Funciona igual que un cronómetro o un conteo en voz alta hasta llegar a la meta.

```python
limite = int(input("Introduzca un número positivo: "))
for i in range(1, limite + 1):
  print(f"Numero: {i}")
```
