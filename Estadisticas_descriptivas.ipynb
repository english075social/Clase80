```python
import numpy as np
import statistics as st
# Pesos de las manzanas en gramos
manzanas = [150, 180, 200, 170, 150, 190, 200, 210, 150, 180]

#Estadisticas descriptivas
media = np.mean(manzanas) # promedio
mediana = np.median(manzanas) # valor central
moda = st.mode(manzanas) # moda
desviacion_estandar = np.std(manzanas) # desviacion estandar / variavilidad

# resultados
print(f"Analisis de la cosecha de manzanas")
print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Moda: {moda}")
print(f"Desviación estándar: {desviacion_estandar:.2f} gramos")
```
