```python
import pandas as pd
# Corregido: pd.DataFrame en lugar de DataGame, eliminada la asignación extra de pd, el símbolo ^ y corregida la mayúscula en manzanas.
datos = pd.DataFrame({"manzanas": [3, 5], "peras": [6, 2]}, index=["Juan", "Pedro"])
datos.plot.bar()
```

```python
datos["Frutas en total"] = datos["manzanas"] + datos["peras"]
# Corregido: .plot.pie en lugar de .plot.py para generar un gráfico de pastel
datos.plot.pie(y="Frutas en total", autopct='%1.1f%%')
```
