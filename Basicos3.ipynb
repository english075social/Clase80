Conversor de Temperatura

```python
grados_centigrados = float(input("Introduce la temperatura en grados centígrados(oC): "))
grados_farhrenheit = (grados_centigrados * 9/5) + 32
print(f"{grados_centigrados} oC equivalen exactamente a {grados_farhrenheit} oF")
```
