Ejercicio 6: Lista de la Compra
(Fundamentos: Estructura de datos list, método .append() y función len())

```python
lista_de_compra = []

lista_de_compra.append("Pan")
lista_de_compra.append("Carne")
lista_de_compra.append("Huevos")
print(f"Lista de la compra actual:{lista_de_compra}")
cantidad_de_elementos = len(lista_de_compra)
print(f"Tienes{cantidad_de_elementos} productos en tu lista. ")
```
